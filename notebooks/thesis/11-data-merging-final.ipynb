```python
import pandas as pd
import numpy as np

# --- File Paths ---
input_file = '../../data/thesis/final_model_dataset_v2.csv'
output_file = '../../data/thesis/final_model_dataset_v3.csv' # New file for the updated dataset

try:
    # Load your final dataset
    df = pd.read_csv(input_file)
    print(f"Loaded dataset with shape: {df.shape}")

    # --- 2. Calculate HOMO-LUMO Gaps ---
    # The gap is always LUMO (higher energy) - HOMO (lower energy)
    df['cation_HOMO_LUMO_Gap_Eh'] = df['cation_LUMO_Eh'] - df['cation_HOMO_Eh']
    df['anion_HOMO_LUMO_Gap_Eh'] = df['anion_LUMO_Eh'] - df['anion_HOMO_Eh']
    
    print("Successfully calculated HOMO-LUMO gaps for cations and anions.")

    # --- 3. Save the new, enriched dataset ---
    df.to_csv(output_file, index=False)
    print(f"New dataset with {df.shape[1]} columns saved to '{output_file}'")
    
    # --- 4. Preview the new columns ---
    print("\n--- Preview of new gap features ---")
    print(df[['cation_HOMO_Eh', 'cation_LUMO_Eh', 'cation_HOMO_LUMO_Gap_Eh', 
              'anion_HOMO_Eh', 'anion_LUMO_Eh', 'anion_HOMO_LUMO_Gap_Eh']].head())

except Exception as e:
    print(f"An error occurred: {e}")
```
